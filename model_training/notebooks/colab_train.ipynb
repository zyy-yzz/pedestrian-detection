## Training (~3-4 hours on T4 GPU)

YOLOv8n + CBAM + C2fGhost + BiFPN + WIoU

```
# 6. Train
import sys, os
sys.path.insert(0, '/content/pedestrian-detection')
os.chdir('/content/pedestrian-detection')
from model_training.train import load_config, build_enhanced_model
from pathlib import Path
from ultralytics import YOLO

cfg = load_config('model_training/configs/train_config.yaml')
cfg['training']['workers'] = 2
cfg['training']['batch_size'] = 16
cfg['training']['device'] = 0
cfg['data']['data_dir'] = 'datasets/LLVIP'

tc = cfg['training']; mc = cfg['model']; ac = cfg['augmentation']
data_dir = Path('datasets/LLVIP')

model = build_enhanced_model(cfg, device='0')
yolo_model = YOLO(model=model)

yolo_model.train(
    data=str(data_dir / 'data.yaml'),
    epochs=tc['epochs'], batch=tc['batch_size'], imgsz=mc['img_size'],
    optimizer=tc['optimizer'], lr0=tc['lr0'], lrf=tc['lrf'],
    momentum=tc['momentum'], weight_decay=tc['weight_decay'],
    warmup_epochs=tc['warmup_epochs'], cos_lr=tc['cos_lr'],
    close_mosaic=tc['close_mosaic'], workers=tc['workers'],
    device=tc['device'], amp=tc['amp'],
    mosaic=ac['mosaic'], mixup=ac['mixup'],
    project='runs/night_ped_detection',
    name=f"enhanced_yolov8n_{tc['epochs']}e",
    exist_ok=True, pretrained=True, verbose=True,
)
print('Training complete!')
```

```
# 7. Save model to Google Drive
import shutil; from pathlib import Path
best = list(Path('runs/night_ped_detection').glob('**/weights/best.pt'))[0]
last = list(Path('runs/night_ped_detection').glob('**/weights/last.pt'))[0]
out = Path('/content/drive/MyDrive/pedestrian-detection/models')
out.mkdir(parents=True, exist_ok=True)
shutil.copy(best, out / 'enhanced_yolov8n.pt')
shutil.copy(last, out / 'enhanced_yolov8n_last.pt')
print(f'Models saved to Drive: {out}')
```

```
# 8. Evaluate
import sys; sys.path.insert(0, '/content/pedestrian-detection')
from model_training.evaluate import evaluate
m = evaluate(
    '/content/drive/MyDrive/pedestrian-detection/models/enhanced_yolov8n.pt',
    'model_training/configs/train_config.yaml', '0')
print(f"mAP@0.5={m['mAP_0.5']:.4f} (target>=0.78) | mAP@0.5:0.95={m['mAP_0.5_0.95']:.4f} (target>=0.52) | FPS={m['fps']:.1f}")
```
